# Get data loaders

```python
%reload_ext autoreload
%autoreload 2

import logging

import yaml

from utilities import get_cifar10_loaders

logging.basicConfig(level=logging.DEBUG, force=True)
logger = logging.getLogger(__name__)

with open("config.yaml", "r") as f:
    try:
        config = yaml.safe_load(f)
    except yaml.YAMLError as exc:
        logger.error(exc)

root = config["data"]["root"]
batch_size = config["hparams"]["batch_size"]
num_workers = config["data"]["num_workers"]

data_loaders = get_cifar10_loaders(root, batch_size, num_workers)
```

```python
%matplotlib inline

import matplotlib.pyplot as plt
import torch

features, labels = next(iter(data_loaders["train"]))
logger.info(f"Features shape: {features.shape}, labels shape: {labels.shape}")
logger.debug(f"Features: mean {features.mean():0.3f}, std {features.std():0.3f}, min {features.min():0.3f}, max {features.max():0.3f}")
logger.debug(f"Labels: {torch.bincount(labels)}")

num_images = 9
num_cols = 3
fig_size_multiplier = 1
num_images = min(num_images, len(features))
num_rows = (num_images + num_cols - 1) // num_cols
fig, axes = plt.subplots(num_rows, num_cols, figsize=(fig_size_multiplier*num_rows, fig_size_multiplier*num_cols))
for ax, feature, label in zip(axes.flatten(), features, labels):
    feature = feature * 0.5 + 0.5
    ax.imshow(feature.permute(1, 2, 0))
    ax.set_title(f"{label}")
    ax.set_axis_off()

fig.tight_layout()
```
