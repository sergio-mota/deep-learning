# Get data loaders

```python
%reload_ext autoreload
%autoreload 2

import logging

import yaml

from data_loaders import get_cifar10_loaders as get_loaders

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

with open("../config/cifar10.yaml", "r") as f:
    try:
        config = yaml.safe_load(f)
    except yaml.YAMLError as exc:
        logger.error(exc)

root = str("../" / Path(config["data"]["root"]))
batch_size = config["hparams"]["batch_size"]
num_workers = config["data"]["num_workers"]

data_loaders = get_loaders(root, batch_size, num_workers)
```

```python
%matplotlib inline

import matplotlib.pyplot as plt
import torch

features, labels = next(iter(data_loaders["train"]))
logger.info(f"Features shape: {features.shape}, labels shape: {labels.shape}")
logger.info(
    f"Features: " 
    f"mean {features.mean():0.3f}, "
    f"std {features.std():0.3f}, "
    f"min {features.min():0.3f}, "
    f"max {features.max():0.3f}"
)
logger.info(f"Labels: {torch.bincount(labels)}")

num_images = 9
num_cols = 3
fig_size_multiplier = 1
num_images = min(num_images, len(features))
num_rows = (num_images + num_cols - 1) // num_cols
fig, axes = plt.subplots(
    num_rows,
    num_cols,
    figsize=(fig_size_multiplier*num_rows, fig_size_multiplier*num_cols)
)
for ax, feature, label in zip(axes.flatten(), features, labels):
    feature = feature * 0.5 + 0.5
    ax.imshow(feature.permute(1, 2, 0))
    ax.set_title(f"{label}")
    ax.set_axis_off()

fig.tight_layout()
```
